## Q2. How many customers have churned, and what is the churn rate?


```python
a= 0
for i in df["Churn"]:
    if i == "Yes":
        a+=1

print("Total Churn :",a)
```

```python
df["Churn"].value_counts()
```

```python
# calculate the rate of customer churn 
churn_customers = 0 
for i in df["Churn"]:
    if i == "Yes":
        churn_customers += 1

print("Churned customers :" , churn_customers)
```

```python
total_customers = df["CustomerID"].nunique()
print("Total Customers :" ,total_customers)
```

```python
# Calculate the rate 

churn_rate = (churn_customers / total_customers)* 100
```

```python
print("Rate of Churn Customers" , round(churn_rate, 2))
```

## Q3. Which contract type has the highest churn rate?

```python
df["Contract"].unique()
```

```python
df.head()
```

```python
 df["Contract"].value_counts()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate churn rate for each contract type
churn_rate = (
    df.groupby("Contract")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="Churn_Rate")
)

# Create bar chart
plt.figure(figsize=(8, 5))
sns.barplot(data=churn_rate, x="Contract", y="Churn_Rate")

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.tight_layout()
plt.show()

```

## Q.6: Does TechSupport affect Customer Churn?

```python
churn_df = df[df["Churn"] == "Yes"]

a = churn_df["TechSupport"].value_counts()

plt.figure(figsize=(8, 6))
plt.pie(
    a,
    labels=a.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.legend(title="Tech Support")
plt.title("Churned Customers by Tech Support")
plt.show()
```

## Q.8: Average Monthly Charges — Churned vs Retained Customers

```python
avg_charges = df.groupby("Churn")["MonthlyCharges"].mean()

avg_charges.plot(kind="bar", figsize=(8, 5))

plt.title("Average Monthly Charges by Churn")
plt.xlabel("Churn")
plt.ylabel("Average Monthly Charges")
plt.xticks(rotation=0)
plt.show()
```

## Q.9: Which customer segment should we target for retention?

```python
churn_rate = (
    df.groupby("Contract")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

churn_rate.plot(kind="bar")

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=15)
plt.show()
```
